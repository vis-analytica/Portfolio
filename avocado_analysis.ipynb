## KPI (Key Performance Indicator)

Es handelt sich um ein Leistungskennzahl, die verwendet wird, um den Erfolg oder Fortschritt in Bezug auf ein spezifisches Ziel zu messen. KPIs sind besonders in Bereichen wie Wirtschaft, Management, Marketing und Datenanalyse wichtig, um die Leistung verschiedener Prozesse, Strategien oder Aktivitäten zu bewerten.

**Eigenschaften eines guten KPI sind:**
- Spezifisch: Der KPI sollte klar und präzise definieren, was gemessen wird.
- Messbar: Der KPI sollte quantifizierbar sein, d.h. er sollte in Zahlen oder Prozentsätzen ausgedrückt werden können.
- Erreichbar: Das Ziel des KPI sollte realistisch und erreichbar sein.
- Relevant: Der KPI sollte in direktem Zusammenhang mit den Zielen des Unternehmens oder Projekts stehen.
- Zeitgebunden: Der KPI sollte an eine bestimmte Zeitspanne gebunden sein, z.B. monatlich, quartalsweise, jährlich.

Angenommen, du bist ein Avocado-Händler und möchtest wissen, wie erfolgreich dein Verkauf in einer bestimmten Region ist. Dann könnten KPIs wie zum Beispiel "Durchschnittlicher Verkaufspreis" entscheidend sein, um den Markttrend und die Nachfrage zu bewerten. Eine signifikante Änderung in diesen KPIs könnte darauf hinweisen, dass Anpassungen an der Preisstrategie oder Marketingkampagnen notwendig sind.
</br>
</br>

**Welche KPIs könnten für uns interessant sein?**
- Durchschnittspreis (Average Price): Der durchschnittliche Verkaufspreis von Avocados in einer bestimmten Region oder in einem bestimmten Zeitraum. Dieser KPI hilft, die Preisentwicklung zu überwachen.
- Gesamtvolumen (Total Volume): Die Gesamtmenge der verkauften Avocados. Dies kann anzeigen, wie gut sich das Produkt in verschiedenen Regionen verkauft und wie stark die Nachfrage ist.
- Verkaufsvolumen nach Typ (Volume by Type): Vergleich des Verkaufsvolumens zwischen konventionellen und biologischen Avocados. Dieser KPI hilft, den Marktanteil und die Beliebtheit der beiden Typen zu verstehen.
- Preis-Volumen-Korrelation: Untersuchung der Beziehung zwischen Preis und Verkaufsvolumen. Ein negativer Zusammenhang kann darauf hindeuten, dass höhere Preise zu geringeren Verkaufszahlen führen.
- Jährliche Wachstumsrate: Die prozentuale Veränderung des Durchschnittspreises oder Verkaufsvolumens im Vergleich zum Vorjahr. Dieser KPI zeigt das Wachstum oder den Rückgang des Marktes über die Jahre.
</br>
</br>

**Warum sind KPIs wichtig?**
- Zielverfolgung: KPIs ermöglichen es, Fortschritte in Richtung festgelegter Ziele zu messen.
- Leistungsbewertung: Sie helfen dabei, die Leistung verschiedener Geschäftsbereiche, Produkte oder Strategien zu bewerten.
- Entscheidungsfindung: KPIs liefern wertvolle Einblicke, die bei der Entscheidungsfindung und strategischen Planung unterstützen.
- Frühwarnsystem: KPIs können frühzeitig auf Probleme hinweisen, sodass rechtzeitig Maßnahmen ergriffen werden können.


Zusammenfassend sind KPIs essentielle Werkzeuge zur Messung und Verbesserung der Leistung in einem Geschäfts- oder Analysekontext. Sie bieten klare, messbare Ziele und helfen, Entscheidungen auf fundierten Daten zu basieren.


**Mögliche KPIs für die Avocado-Datenanalyse**

1. Durchschnittlicher Verkaufspreis (Average Price):
   1. Beschreibung: Durchschnittlicher Preis pro Avocado (konventionell und biologisch).
   2. Fragestellung: Wie entwickeln sich die durchschnittlichen Preise über die Jahre und in verschiedenen Regionen?
   3. Ziel: Preisentwicklung und Preisstabilität überwachen, um Markttrends zu verstehen.
2. Gesamtverkaufsvolumen (Total Volume):
   1. Beschreibung: Die gesamte Anzahl verkaufter Avocados in einer bestimmten Region oder einem bestimmten Zeitraum.
   2. Fragestellung: Wie viel Volumen wird in den verschiedenen Regionen verkauft? Gibt es Spitzen in bestimmten Jahreszeiten?
   3. Ziel: Nachfrage und Saisonalität identifizieren.
3. Verkaufsvolumen nach Typ (Volume by Type):
   1. Beschreibung: Vergleich des Verkaufsvolumens zwischen konventionellen und biologischen Avocados.
   2. Fragestellung: Welche Art von Avocados ist beliebter (konventionell oder biologisch) und wie variiert dies in verschiedenen Regionen?
   3. Ziel: Marktanteile der verschiedenen Produktkategorien bewerten.
4. Umsatz pro Region (Revenue per Region):
   1. Beschreibung: Gesamtumsatz in einer bestimmten Region (Durchschnittspreis * Gesamtvolumen).
   2. Fragestellung: Welche Regionen generieren den höchsten Umsatz?
   3. Ziel: Regionen mit hoher Verkaufsleistung identifizieren und Marketingmaßnahmen gezielt einsetzen.
5. Jährliche Wachstumsrate (Annual Growth Rate):
   1. Beschreibung: Prozentuale Änderung des Verkaufsvolumens und des Preises im Vergleich zum Vorjahr.
   2. Fragestellung: Gibt es ein Wachstum oder einen Rückgang im Verkauf und Preis? Welche Regionen wachsen am schnellsten?
   3. Ziel: Langfristige Wachstums- und Rückgangstrends überwachen.
6. Preis-Volumen-Korrelation (Price-Volume Correlation):
   1. Beschreibung: Beziehung zwischen dem Preis und dem Verkaufsvolumen.
   2. Fragestellung: Gibt es eine negative Korrelation, die darauf hinweist, dass höhere Preise zu geringeren Verkaufszahlen führen?
   3. Ziel: Verständnis des Einflusses von Preisänderungen auf die Verkaufszahlen.
7. Saisonale Nachfrage (Seasonal Demand):
   1. Beschreibung: Veränderung des Verkaufsvolumens und Preises über das Jahr hinweg.
   2. Fragestellung: Gibt es saisonale Spitzen, z.B. zur Weihnachtszeit oder im Sommer?
   3. Ziel: Bessere Planung der Lagerbestände und Marketingaktionen basierend auf saisonaler Nachfrage.
8. Marktanteil (Market Share):
   1. Beschreibung: Anteil des Verkaufsvolumens einer Region im Vergleich zum Gesamtvolumen.
   2. Fragestellung: Welche Region hat den größten Marktanteil? Gibt es Veränderungen im Marktanteil über die Jahre?
   3. Ziel: Wettbewerbsfähigkeit in verschiedenen Regionen bewerten.

**Mögliche Fragestellungen für eine tiefere Analyse:**
1. Preisentwicklung: Wie hat sich der Preis für Avocados in den letzten Jahren entwickelt? Gibt es Unterschiede in der Preisentwicklung zwischen konventionellen und biologischen Avocados?
2. Regionale Unterschiede: Welche Regionen zeigen die höchste Nachfrage nach Avocados? Welche Regionen haben die höchsten Preise?
3. Saisonale Trends: Gibt es bestimmte Jahreszeiten, in denen der Avocado-Verkauf signifikant ansteigt oder abnimmt?
4. Verhältnis von Preis und Volumen: Wie beeinflusst eine Preisänderung das Verkaufsvolumen? Gibt es einen optimalen Preis, bei dem das Volumen maximiert wird?
5. Nachfrage nach Bioprodukten: Wie hat sich die Nachfrage nach biologischen Avocados im Vergleich zu konventionellen entwickelt? Welche Faktoren könnten diesen Unterschied erklären?
6. Umsatzentwicklung pro Jahr: Wie entwickelt sich der Umsatz insgesamt und in den einzelnen Regionen? Welche Regionen sind für das größte Wachstum verantwortlich?

### 4.4 Umsatz pro Region (Revenue per Region):

Der Umsatz wird berechnet, indem der `AveragePrice` mit dem `Total Volume` multipliziert wird.

```python
# Erstellung einer Kopie des DataFrames
avocado_data = avocado_data.copy()

# Berechnung des Umsatzes für jede Zeile
avocado_data['Revenue'] = avocado_data['AveragePrice'] * avocado_data['Total Volume']
avocado_data.head()
```

Nun können wir den Gesamtumsatz pro Region berechnen, indem wir die Revenue-Spalte nach der region-Spalte gruppieren und summieren:

```python
# Gruppierung nach Region und Berechnung des Gesamtumsatzes
revenue_per_region = avocado_data.groupby('region')['Revenue'].sum().reset_index()

# Anzeige der ersten Einträge der Gruppierung
print("Umsatz pro Region:")
revenue_per_region.head()
```

```python
# Absteigend sortieren
revenue_per_region.sort_values(by="Revenue", ascending=False).head(10)
```

**Interpretation:**

- Marktgröße und Verteilung:
  - `TotalUS` gibt den Gesamtrahmen des nationalen Marktes vor. Regionen wie `California`, `West` und `Northeast` tragen erheblich zu diesem Gesamtumsatz bei. Diese Regionen haben eine starke Nachfrage und sind für den Avocado-Markt von entscheidender Bedeutung.
- Hoher Umsatz in Kalifornien:
  - Kalifornien, als Hauptproduzent von Avocados, zeigt einen sehr hohen Umsatz. Dies spiegelt sowohl die Produktion als auch den Konsum innerhalb des Bundesstaates wider.
- Starke Nachfrage im Nordosten und Südosten:
  - Regionen wie `Northeast` und `Southeast` haben ebenfalls starke Märkte, was auf eine wachsende Beliebtheit von Avocados in städtischen und dicht besiedelten Gebieten hinweist.
- Wachstumschancen in kleineren Märkten:
  - Regionen wie `Plains` und `Midsouth` haben kleinere Marktanteile, könnten aber potenziell weiter wachsen, wenn die Nachfrage dort steigt oder gezielte Marketingmaßnahmen umgesetzt werden.

### 4.5 Preis-Volumen-Korrelation (Price-Volume Correlation):

Diese Korrelation hilft uns zu verstehen, ob es einen Zusammenhang zwischen dem Preis von Avocados und der Menge der verkauften Avocados gibt. Eine negative Korrelation könnte darauf hinweisen, dass höhere Preise zu geringeren Verkaufsvolumen führen, während eine positive Korrelation das Gegenteil bedeutet.

Der Korrelationskoeffizient kann zwischen -1 und 1 liegen:

- +1: Perfekte positive Korrelation (wenn der Preis steigt, steigt auch das Volumen).
- -1: Perfekte negative Korrelation (wenn der Preis steigt, sinkt das Volumen).
- 0: Keine Korrelation (kein Zusammenhang zwischen Preis und Volumen).

```python
# Berechnung der Korrelation zwischen AveragePrice und Total Volume
price_volume_correlation = avocado_data['AveragePrice'].corr(avocado_data['Total Volume'])

print("Korrelation zwischen Durchschnittspreis und Gesamtvolumen:")
price_volume_correlation
```

**Interpretation:**

- Da der Wert negativ ist, deutet dies darauf hin, dass tendenziell ein leichter negativer Zusammenhang zwischen Preis und Volumen besteht. Das bedeutet:
  - Wenn der Preis steigt, nimmt das Verkaufsvolumen leicht ab
  - Wenn der Preis sinkt, nimmt das Verkaufsvolumen leicht zu.
- Schwache Korrelation (nahe 0):
  - Der Wert liegt nahe bei 0, was auf eine schwache Korrelation hindeutet. Das bedeutet, dass der Zusammenhang zwischen Preis und Volumen nicht stark ist.
  - In praktischen Worten bedeutet dies, dass der Preis nur einen geringen Einfluss auf die verkaufte Menge hat. Andere Faktoren wie Saisonalität, Verfügbarkeit, Nachfrage und Marketing könnten ebenfalls eine bedeutende Rolle spielen.

### Zusatz: Jährliche Wachstumsrate (Annual Growth Rate):

```python
# Schritt 1: Gruppieren nach Jahr und Berechnen des jährlichen Durchschnittspreises und Gesamtvolumens
annual_data = avocado_data.groupby('year').agg({
    'AveragePrice': 'mean',   # Durchschnittlicher Preis pro Jahr
    'Total Volume': 'sum'     # Gesamtvolumen pro Jahr
}).reset_index()

# Initialisieren der neuen Spalten für die Wachstumsraten
annual_data['Price Growth Rate'] = 0.0
annual_data['Volume Growth Rate'] = 0.0

# Schritt 2: Berechnung der Wachstumsraten für jedes Jahr, ab dem zweiten Jahr
for i in range(1, len(annual_data)):
    # Berechnung der Wachstumsrate des Durchschnittspreises
    previous_price = annual_data.loc[i - 1, 'AveragePrice']
    current_price = annual_data.loc[i, 'AveragePrice']
    if previous_price != 0:  # Vermeidung von Division durch 0
        annual_data.loc[i, 'Price Growth Rate'] = ((current_price - previous_price) / previous_price) * 100

    # Berechnung der Wachstumsrate des Verkaufsvolumens
    previous_volume = annual_data.loc[i - 1, 'Total Volume']
    current_volume = annual_data.loc[i, 'Total Volume']
    if previous_volume != 0:  # Vermeidung von Division durch 0
        annual_data.loc[i, 'Volume Growth Rate'] = ((current_volume - previous_volume) / previous_volume) * 100

# Schritt 3: Ausgabe der berechneten Wachstumsraten
print("Jährliche Wachstumsrate für Durchschnittspreis und Verkaufsvolumen:")
annual_data
```
